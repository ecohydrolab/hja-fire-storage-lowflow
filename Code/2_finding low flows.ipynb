```python
#read in daily Q data
dailyQ=pd.read_csv(daily_flow_data)
dailyQ["DATE"]=pd.to_datetime(dailyQ["DATE"])
# remove October and November from data
dailyQ_nofall=dailyQ[dailyQ['month']!=10]
print(dailyQ_nofall)
```

```python
#group by Wateryear and find 5th percentile flow
def percentile(x):
       return x.quantile(0.05)
wy_lowflow=dailyQ_nofall.groupby("WATERYEAR").agg(GSLOOK=("GSLOOK",percentile),
                                                  GSWSMC=("GSWSMC",percentile),
                                                  GSWS01=("GSWS01",percentile),
                                                  GSWS02=("GSWS02",percentile),
                                                  GSWS03=("GSWS03",percentile),
                                                  GSWS08=("GSWS08",percentile),
                                                  GSWS09=("GSWS09",percentile),
                                                  GSWS10=("GSWS10",percentile)).reset_index()
print(wy_lowflow)
```
